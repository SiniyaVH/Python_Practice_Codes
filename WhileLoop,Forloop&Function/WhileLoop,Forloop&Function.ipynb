```python
import random
secret_number = random.randint(1, 10)
attempt=1
while attempt<=3:
    try:
        guess=int(input("Guess a number between (1 to 10)include 1 and 10:"))
        if 1<=guess<=10:
            if guess==secret_number:
                 print("Excellent!!! Your guessed number is correct")
                 break
            else:
                 print("Ohh!!! No!!  Wrong Guess!!!")
                 attempt+=1
                 if(guess<secret_number):
                     print("Try a higher number")
                 elif(guess>secret_number):
                     print("Try a lower number")
                 continue
        else:
              print("Invalid entry!!! Please enter a value from 1 to 10")
              continue
    except ValueError:
        print("Invalid input!!! Please enter a number.")
else:
    print("your attempt is over")
```
