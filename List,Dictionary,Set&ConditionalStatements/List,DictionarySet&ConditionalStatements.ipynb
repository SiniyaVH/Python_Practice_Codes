## Operators & Conditional Statements :
### (IF, ELIF, ELSE)

##### Prompt user for Input. Score range should be from 0 to 10 (both inclusive)

```python
score=int(input("Enter the score(0-10):"))
if 0<=score and score<=10:
    print("The score is:",score)
else:
    print("Invalid score! Plase enter a score between 0 and 10.")
```

##### Find the performance category based on the input score using following criteria:
##### a. Above Average: Score greater than 7
##### b . Average: Score between 4 and 7(both inclusive)
##### c. Below Average: Score lesser than 4

```python
score=int(input("Enter the score(0-10):"))
if 0<=score and score<=10:
    if score>7:
        print("The Score is Above Average")
    elif 4<=score<=7:
        print("The Score is Average")
    else:
        print("The Score is Below Average")    
else:
    print("Invalid score! Plase enter a score between 0 and 10.")
```

##### Output: Print the Performance category

```python
score=int(input("Enter the score(0-10):"))
if 0<=score and score<=10:
    if score>7 and score<=10:
        print(" Performance category:","Above Average")
    elif 4<=score<=7:
        print(" Performance category:","Average")
    else:
        print(" Performance category:","Below Average")
else:
    print("Invalid score! Plase enter a score between 0 and 10.")
```

##### Additional Step: You can give a prompt of your choice to each category.
##### For eg: If score below average “Need to Improve your performance, consistent practice will lead to better results”

```python
score=int(input("Enter the score(0-10):"))
if 0<=score and score<=10:
    if score>7 and score<=10:
        print(" Performance category:","Above Average")
        print("Excellent performance! Keep up the great work")
    elif 4<=score<=7:
        print(" Performance category:","Average")
        print("Good effort! Keep practicing and you can improve your performance further")
    else:
        print(" Performance category:","Below Average")
        print("Your score is below average. Need to Improve your performance, consistent practice will lead to better results")
else:
    print("Invalid score! Plase enter a score between 0 and 10.")
```
